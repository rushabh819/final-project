```python
import pandas as pd
import matplotlib.pyplot as plt

# Load processed data from the uploaded location
df = pd.read_csv("../data/processed/country_health_econ.csv")

# Plot 1: GDP per Capita over Time for Selected Countries
countries = ['United States', 'India', 'Brazil']
df_gdp_ts = df[df['Country Name'].isin(countries)]

plt.figure()
for country in countries:
    subset = df_gdp_ts[df_gdp_ts['Country Name'] == country]
    plt.plot(subset['Year'], subset['GDP_per_capita'], label=country)
plt.xlabel('Year')
plt.ylabel('GDP per Capita (current US$)')
plt.title('GDP per Capita Over Time')
plt.legend()
plt.tight_layout()
plt.show()

# Plot 2: Time series for selected countries
countries = ['United States', 'India', 'Brazil']
df_timeseries = df[df['Country Name'].isin(countries)]

plt.figure()
for country in countries:
    subset = df_timeseries[df_timeseries['Country Name'] == country]
    plt.plot(subset['Year'], subset['Life_Expectancy'], label=country)
plt.xlabel('Year')
plt.ylabel('Life Expectancy (years)')
plt.title('Life Expectancy Over Time')
plt.legend()
plt.tight_layout()
plt.show()

# Plot 3: Time Series of GDP per Capita & Life Expectancy (Combined Plot)
countries = ['United States', 'India', 'Brazil']
plt.figure()
for country in countries:
    subset = df[df['Country Name'] == country]
    plt.plot(subset['Year'], subset['GDP_per_capita'], label=f'GDP: {country}')
    plt.plot(subset['Year'], subset['Life_Expectancy'], linestyle='--', label=f'LifeExp: {country}')
plt.xlabel('Year')
plt.ylabel('Value')
plt.title('GDP per Capita and Life Expectancy Over Time')
plt.legend(loc='upper left')
plt.tight_layout()
plt.show()
```

## Plot 1: GDP per Capita Over Time
- United States

    - Starts around $3 000–4 000 in the early 1960s, rises steadily through the ’70s and ’80s.
    - Explodes after 1990, reaching over $60 000 by the 2010s.
    - Shows a slight dip around 2020 (COVID-19 downturn) and then rebounds.

- Brazil

    - Much lower base, roughly $300 in 1960.
    - Gradual growth through the ’80s, followed by a sharper climb in the 2000s to around $12 000–14 000.
    - Fluctuates more visibly around global crises (e.g., early-’80s debt crisis, 2015–16 recession).

- India

    - Begins near $100 in 1960, almost flat until the early 1980s.
    - Picks up pace after liberalization in the 1990s, rising to about $2 000–3 000 by the late 2010s.
    - Still far below Brazil and the U.S., but the steepest percentage growth in recent decades.

*Takeaway*: All three economies grow over time, but the scale and rate differ dramatically. The U.S. enjoys both a higher starting point and sustained exponential growth; Brazil and India show more modest gains punctuated by regional or global shocks.

## Plot 2: Life Expectancy Over Time
- United States

    - Begins around 70 years in 1960, steadily climbs to about 78–79 years by the 2010s.
    - Plateaus somewhat in the 2010s, with a small dip around 2020 (pandemic effects).

- Brazil

    - Starts around 53 years in 1960, then a smooth upward trajectory to roughly 75 years by 2020.
    - Gains in healthcare and sanitation drive much of that improvement.

- India

    - Very low baseline near 46 years in 1960.
    - Gradual but accelerating increase through public health campaigns and economic development, reaching around 69–70 years by 2020.

*Takeaway*: All three countries see significant life‐expectancy gains, but India and Brazil achieve the largest absolute improvements (25+ years), while the U.S. gains are more modest (8–9 years) since it started at a higher baseline.

## Plot 3: Combined GDP & Life Expectancy Over Time
Overlaying both indicators for each country highlights their intertwined paths:

- United States (solid gold for GDP, dashed for life expectancy)

    - A dramatic GDP climb accompanies a slower, tapering rise in life expectancy—reflecting diminishing returns of wealth on health at high income levels.

- Brazil (solid red / dashed red)

    - Parallel upward trends: GDP jumps in the 2000s align with surges in life expectancy, showing how economic gains directly translated into health improvements.

- India (solid orange / dashed orange)

    - A more staggered pattern: life expectancy steadily climbs even when GDP growth is more muted (pre-1990), indicating that public health measures can raise lifespan ahead of broad economic development.

#### Overall Insight:

- *Temporal lead/lag*: In India, health gains precede big economic gains; in Brazil and the U.S., economic growth and health improvements are more tightly synchronized.

- *Magnitude of effect*: The U.S.’s huge GDP gains yield comparatively small life-expectancy improvements, whereas middle-income countries see a stronger “bang for the buck” in health outcomes as their economies develop.
